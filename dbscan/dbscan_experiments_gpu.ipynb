## Part 2: Data Loading

Loads embeddings, handles duplicates, computes k-distance for eps estimation.


````python
# Part 2 — Smart file loading with duplicate analysis and k-distance plot

def detect_embedding_dim(file_path: Path) -> int:
    """Auto-detect embedding dimension from filename"""
    filename = file_path.name.lower()
    if 'pca256' in filename:
        return 256
    elif 'pca128' in filename:
        return 128
    else:
        return 768

def infer_num_rows(path: Path, embedding_dim: int = None) -> int:
    """Infer number of rows for RAW memmap files"""
    if embedding_dim is None:
        embedding_dim = detect_embedding_dim(path)
    size = path.stat().st_size
    return size // (embedding_dim * np.dtype(np.float32).itemsize)

def load_single_file_smart(file_path: Path, embedding_dim: int = None):
    """Smart loader: auto-detect .npy vs RAW memmap"""
    if embedding_dim is None:
        embedding_dim = detect_embedding_dim(file_path)
        print(f"   🔍 Auto-detected dimension: {embedding_dim}")
    
    try:
        arr = np.load(file_path, mmap_mode='r')
        detected_dim = arr.shape[1]
        if detected_dim != embedding_dim:
            print(f"   ⚠️ Dimension mismatch! Expected {embedding_dim}, got {detected_dim}")
            embedding_dim = detected_dim
        return arr, True, arr.shape[0]
    except Exception:
        num_rows = infer_num_rows(file_path, embedding_dim)
        arr = np.memmap(file_path, dtype=np.float32, mode='r', 
                       shape=(num_rows, embedding_dim))
        print(f"   ⚠️ Loaded as RAW memmap: {num_rows:,} rows × {embedding_dim} dims")
        return arr, True, num_rows

def load_embeddings_from_files(files):
    """Load embeddings with auto-dimension detection"""
    if len(files) == 0:
        raise FileNotFoundError('No embedding files provided')
    
    for f in files:
        if not f.exists():
            raise FileNotFoundError(f'File not found: {f}')
    
    first_arr, _, _ = load_single_file_smart(files[0])
    embedding_dim = first_arr.shape[1]
    print(f"Embedding dimension: {embedding_dim}")
    
    if len(files) == 1:
        print(f"Loading single file: {files[0].name}")
        return first_arr
    
    print(f"Loading {len(files)} files...")
    total_samples = 0
    file_info = []
    for f in files:
        arr, is_mmap, n_rows = load_single_file_smart(f)
        file_info.append((f, arr, n_rows))
        total_samples += n_rows
        print(f"  - {f.name}: {n_rows:,} rows")
    
    total_size_gb = (total_samples * embedding_dim * 4) / (1024**3)
    print(f"\nTotal samples: {total_samples:,} ({total_size_gb:.2f} GB)")
    
    if total_size_gb < 100:
        arrays = [arr for _, arr, _ in file_info]
        return np.vstack(arrays)
    else:
        raise MemoryError(f"Dataset too large ({total_size_gb:.1f}GB)")

# Load embeddings
print("Loading embeddings...")
emb = load_embeddings_from_files(INPUT_FILES)
print(f'Loaded embeddings shape: {emb.shape}')

# Normalize for cosine distance
if USE_COSINE_DISTANCE:
    print('\n🔄 Normalizing embeddings for cosine distance...')
    emb = normalize(emb, norm='l2')
    print(f'✓ Normalized')

# Duplicate analysis
print('\n🔍 Analyzing duplicates...')
n_total = emb.shape[0]
rng = np.random.RandomState(RANDOM_STATE)

if n_total > SAMPLE_FOR_KDIST:
    dup_check_idx = rng.choice(n_total, min(SAMPLE_FOR_KDIST, n_total), replace=False)
    emb_dup_check = emb[dup_check_idx]
else:
    emb_dup_check = emb
    dup_check_idx = None

unique_embeddings = np.unique(emb_dup_check, axis=0)
duplicate_ratio = 1 - (len(unique_embeddings) / len(emb_dup_check))

print(f'   Duplicate ratio: {duplicate_ratio:.2%}')

use_unique_for_kdist = HANDLE_DUPLICATES and duplicate_ratio > DUPLICATE_THRESHOLD

if use_unique_for_kdist:
    print(f'   → High duplicates, using unique samples for k-distance')
    if n_total > SAMPLE_FOR_KDIST * 2:
        sample_idx = rng.choice(n_total, SAMPLE_FOR_KDIST, replace=False)
        unique_for_kdist = np.unique(emb[sample_idx], axis=0)
    else:
        unique_for_kdist = np.unique(emb, axis=0)
    emb_for_kdist = unique_for_kdist
else:
    print(f'   → Low duplicates, standard sampling')
    if n_total > SAMPLE_FOR_KDIST:
        kdist_idx = rng.choice(n_total, SAMPLE_FOR_KDIST, replace=False)
        emb_for_kdist = emb[kdist_idx]
    else:
        emb_for_kdist = emb

del unique_embeddings
if dup_check_idx is not None:
    del emb_dup_check
gc.collect()

# Compute k-distance
print(f'\n🔄 Computing k-distance (k={KNN_NEIGHBORS})...')
print(f'   Sample size: {len(emb_for_kdist):,}')

nn = NearestNeighbors(n_neighbors=KNN_NEIGHBORS, n_jobs=-1, metric='euclidean')
nn.fit(emb_for_kdist)
distances, _ = nn.kneighbors(emb_for_kdist)
k_dist = np.sort(distances[:, -1])

del nn, distances
gc.collect()

# Filter zeros
k_dist_nonzero = k_dist[k_dist > 0]
if len(k_dist_nonzero) < len(k_dist):
    pct_zero = (1 - len(k_dist_nonzero) / len(k_dist)) * 100
    print(f'   ⚠️ {pct_zero:.1f}% distances are 0 (duplicates)')
    k_dist_for_stats = k_dist_nonzero
else:
    k_dist_for_stats = k_dist

# Plot k-distance
plt.figure(figsize=(10,4))
plt.plot(k_dist)
plt.xlabel('Points sorted by k-distance')
plt.ylabel(f'k-distance (k={KNN_NEIGHBORS})')
plt.title('k-distance plot — look for elbow to choose eps')
if len(k_dist_for_stats) > 0:
    plt.axhline(y=np.percentile(k_dist_for_stats, 90), color='r', linestyle='--', alpha=0.5, label='90th percentile')
    plt.axhline(y=np.percentile(k_dist_for_stats, 95), color='orange', linestyle='--', alpha=0.5, label='95th percentile')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f'\n📊 k-distance statistics:')
if len(k_dist_for_stats) > 0:
    for p in [50, 60, 70, 80, 85, 90, 95]:
        print(f'   {p:2d}th percentile: {np.percentile(k_dist_for_stats, p):.4f}')

k_dist_clean = k_dist_for_stats
del k_dist, k_dist_nonzero
if 'emb_for_kdist' in locals() and emb_for_kdist is not emb:
    del emb_for_kdist
gc.collect()

print(f'\n✅ Part 2 complete')
````

## Part 3: Parameter Optimization

- 5-10x faster per iteration
- Total time GPU: 2-5 min vs CPU: 10-20 min


````python
# Part 3 — GPU-Accelerated Parameter Grid Search

print('Preparing sample for parameter search...')
n = emb.shape[0]

if n > SAMPLE_FOR_METRICS:
    rng = np.random.RandomState(RANDOM_STATE)
    sample_idx = rng.choice(n, SAMPLE_FOR_METRICS, replace=False)
    emb_sample = emb[sample_idx].copy()
    print(f'Sample: {SAMPLE_FOR_METRICS:,} / {n:,}')
    del sample_idx
    gc.collect()
else:
    emb_sample = emb
    print(f'Using full dataset: {n:,}')

# Parameter grid (GPU can afford larger grid)
if len(k_dist_clean) > 0:
    eps_min = np.percentile(k_dist_clean, 75)
    eps_max = np.percentile(k_dist_clean, 99)
    if eps_max - eps_min < 0.01:
        eps_min = np.percentile(k_dist_clean, 50)
    n_eps = 8 if USE_GPU else 6
    eps_values = np.linspace(eps_min, eps_max, n_eps)
else:
    eps_values = np.linspace(0.1, 2.0, 6)

min_samples_values = [5, 10, 15, 20, 30] if USE_GPU else [10, 20, 30]

print(f'\nParameter grid:')
print(f'  eps: {len(eps_values)} values [{eps_values.min():.4f} - {eps_values.max():.4f}]')
print(f'  min_samples: {min_samples_values}')
print(f'  Total: {len(eps_values) * len(min_samples_values)} combinations')
print(f'  Algorithm: {"🎮 cuML DBSCAN (GPU)" if USE_GPU else "💻 sklearn DBSCAN (CPU)"}')

results = []
pbar = tqdm(total=len(eps_values) * len(min_samples_values), 
            desc='Grid search', unit='config')

if USE_GPU:
    # GPU path
    import cupy as cp
    emb_sample_gpu = cp.asarray(emb_sample, dtype=cp.float32)
    
    for eps in eps_values:
        for ms in min_samples_values:
            start_time = time.time()
            
            dbs = cuDBSCAN(eps=float(eps), min_samples=int(ms))
            labels = dbs.fit_predict(emb_sample_gpu)
            labels_cpu = cp.asnumpy(labels)
            
            n_clusters = len(set(labels_cpu) - {-1})
            n_noise = np.sum(labels_cpu == -1)
            noise_pct = (n_noise / len(labels_cpu)) * 100
            
            sil = -1
            if n_clusters > 1:
                try:
                    sil = silhouette_score(emb_sample, labels_cpu, n_jobs=-1)
                except:
                    pass
            
            results.append({
                'eps': float(eps),
                'min_samples': int(ms),
                'n_clusters': n_clusters,
                'noise_pct': float(noise_pct),
                'silhouette': float(sil)
            })
            
            elapsed = time.time() - start_time
            pbar.update(1)
            tqdm.write(f'  eps={eps:.4g}, ms={ms:2d} → clusters={n_clusters:2d}, noise={noise_pct:5.1f}%, sil={sil:6.4f} ({elapsed:.2f}s)')
            
            del dbs, labels, labels_cpu
            gc.collect()
    
    del emb_sample_gpu
    cp.get_default_memory_pool().free_all_blocks()
    
else:
    # CPU path
    for eps in eps_values:
        for ms in min_samples_values:
            start_time = time.time()
            
            dbs = DBSCAN(eps=float(eps), min_samples=int(ms), 
                        algorithm='ball_tree', n_jobs=-1)
            labels = dbs.fit_predict(emb_sample)
            
            n_clusters = len(set(labels) - {-1})
            n_noise = np.sum(labels == -1)
            noise_pct = (n_noise / len(labels)) * 100
            
            sil = -1
            if n_clusters > 1:
                try:
                    sil = silhouette_score(emb_sample, labels, n_jobs=-1)
                except:
                    pass
            
            results.append({
                'eps': float(eps),
                'min_samples': int(ms),
                'n_clusters': n_clusters,
                'noise_pct': float(noise_pct),
                'silhouette': float(sil)
            })
            
            elapsed = time.time() - start_time
            pbar.update(1)
            tqdm.write(f'  eps={eps:.4g}, ms={ms:2d} → clusters={n_clusters:2d}, noise={noise_pct:5.1f}%, sil={sil:6.4f} ({elapsed:.1f}s)')
            
            del dbs, labels
            gc.collect()

pbar.close()
del emb_sample
gc.collect()

# Results
df_res = pd.DataFrame(results)
print('\n' + '='*70)
print('TOP 10 CONFIGURATIONS (by silhouette)')
print('='*70)
print(df_res.sort_values('silhouette', ascending=False).head(10).to_string(index=False))

print('\n' + '='*70)
print('CONFIGURATIONS WITH LOW NOISE (<30%)')
print('='*70)
df_low_noise = df_res[df_res['noise_pct'] < 30].sort_values('silhouette', ascending=False)
if len(df_low_noise) > 0:
    print(df_low_noise.head(10).to_string(index=False))
else:
    print('⚠️ No configs with <30% noise found')

print('\n✅ Part 3 complete')
````

## Part 4: Final DBSCAN Clustering

**GPU Performance:**
- BGL 4M (768d): ~5-10 minutes
- BGL 4M (PCA256): ~2-5 minutes  
- Thunderbird 200M (PCA128): ~1-2 hours (with batching)

**CPU Performance (comparison):**
- BGL 4M: ~30-60 minutes
- Thunderbird 200M: ❌ Impossible


````python
# Part 4 — GPU-Accelerated Final DBSCAN Clustering

best_config = df_res.sort_values('silhouette', ascending=False).iloc[0]

CHOSEN_EPS = float(best_config['eps'])
CHOSEN_MIN_SAMPLES = int(best_config['min_samples'])

print('='*70)
print('FINAL DBSCAN CONFIGURATION')
print('='*70)
print(f'Algorithm:   {"🎮 cuML DBSCAN (GPU)" if USE_GPU else "💻 sklearn DBSCAN (CPU)"}')
print(f'eps:         {CHOSEN_EPS:.6f}')
print(f'min_samples: {CHOSEN_MIN_SAMPLES}')
print(f'metric:      euclidean' + (' (on normalized = cosine)' if USE_COSINE_DISTANCE else ''))
print(f'Expected clusters: {int(best_config["n_clusters"])}')
print(f'Expected noise:    {best_config["noise_pct"]:.1f}%')
print(f'Expected silhouette: {best_config["silhouette"]:.4f}')
print('='*70)

# Cleanup
for var in ['df_res', 'results', 'df_low_noise', 'k_dist_clean']:
    if var in locals():
        del locals()[var]
gc.collect()

n_total = emb.shape[0]
emb_size_gb = emb.nbytes / (1024**3)

if USE_GPU:
    # ========================================================================
    # GPU CLUSTERING PATH
    # ========================================================================
    import cupy as cp
    
    gpu_props = cp.cuda.runtime.getDeviceProperties(0)
    available_vram_gb = gpu_props['totalGlobalMem'] / (1024**3)
    usable_vram_gb = available_vram_gb * 0.7  # Use 70% for safety
    max_samples_per_batch = int((usable_vram_gb * (1024**3)) / (emb.shape[1] * 4))
    
    print(f'\n📊 GPU Dataset Analysis:')
    print(f'   Total samples: {n_total:,}')
    print(f'   Dataset size: {emb_size_gb:.2f} GB')
    print(f'   GPU VRAM: {available_vram_gb:.1f} GB (usable: {usable_vram_gb:.1f} GB)')
    print(f'   Max batch size: {max_samples_per_batch:,} samples')
    
    if n_total <= max_samples_per_batch:
        # Single GPU pass
        print(f'\n🚀 GPU Single-Pass Clustering')
        print(f'   Processing {n_total:,} samples in one batch')
        print(f'   Expected time: 5-15 minutes')
        
        start_time = time.time()
        
        print(f'   → Transferring to GPU...')
        emb_gpu = cp.asarray(emb, dtype=cp.float32)
        
        print(f'   → Fitting DBSCAN on GPU...')
        model = cuDBSCAN(eps=CHOSEN_EPS, min_samples=CHOSEN_MIN_SAMPLES)
        labels_gpu = model.fit_predict(emb_gpu)
        labels_full = cp.asnumpy(labels_gpu)
        
        elapsed = time.time() - start_time
        print(f'   ✅ GPU DBSCAN completed in {elapsed/60:.1f} minutes!')
        
        del emb_gpu, labels_gpu, model
        cp.get_default_memory_pool().free_all_blocks()
        gc.collect()
        
    else:
        # Batch processing
        n_batches = (n_total + max_samples_per_batch - 1) // max_samples_per_batch
        print(f'\n🚀 GPU Batch Processing')
        print(f'   {n_batches} batches of ~{max_samples_per_batch:,} samples')
        print(f'   Strategy: Fit on representative sample → k-NN assignment')
        
        # Fit on merged sample from all batches
        samples_per_batch = max(10000, max_samples_per_batch // n_batches)
        fit_indices = []
        
        for batch_idx in range(n_batches):
            start_idx = batch_idx * max_samples_per_batch
            end_idx = min((batch_idx + 1) * max_samples_per_batch, n_total)
            batch_size = end_idx - start_idx
            
            n_sample = min(samples_per_batch, batch_size)
            rng = np.random.RandomState(RANDOM_STATE + batch_idx)
            batch_indices = rng.choice(batch_size, n_sample, replace=False) + start_idx
            fit_indices.extend(batch_indices)
        
        fit_indices = np.array(fit_indices)
        print(f'\n   Step 1: Fitting on {len(fit_indices):,} samples')
        
        emb_fit = emb[fit_indices]
        emb_fit_gpu = cp.asarray(emb_fit, dtype=cp.float32)
        
        start_time = time.time()
        model = cuDBSCAN(eps=CHOSEN_EPS, min_samples=CHOSEN_MIN_SAMPLES)
        labels_fit_gpu = model.fit_predict(emb_fit_gpu)
        labels_fit = cp.asnumpy(labels_fit_gpu)
        
        elapsed = time.time() - start_time
        print(f'   ✅ Fit completed in {elapsed/60:.1f} minutes')
        print(f'   Found {len(set(labels_fit) - {-1})} clusters')
        
        del labels_fit_gpu, model
        cp.get_default_memory_pool().free_all_blocks()
        
        # k-NN assignment
        remaining_idx = np.setdiff1d(np.arange(n_total), fit_indices)
        print(f'\n   Step 2: Assigning {len(remaining_idx):,} points via GPU k-NN')
        
        labels_full = np.zeros(n_total, dtype=np.int32)
        labels_full[fit_indices] = labels_fit
        
        from cuml.neighbors import NearestNeighbors as cuNearestNeighbors
        knn = cuNearestNeighbors(n_neighbors=5)
        knn.fit(emb_fit_gpu)
        
        ASSIGN_BATCH = 500000
        n_assign_batches = (len(remaining_idx) + ASSIGN_BATCH - 1) // ASSIGN_BATCH
        
        for batch_idx in tqdm(range(n_assign_batches), desc='GPU k-NN', unit='batch'):
            start_idx = batch_idx * ASSIGN_BATCH
            end_idx = min((batch_idx + 1) * ASSIGN_BATCH, len(remaining_idx))
            batch_idx_arr = remaining_idx[start_idx:end_idx]
            
            emb_batch_gpu = cp.asarray(emb[batch_idx_arr], dtype=cp.float32)
            distances, indices = knn.kneighbors(emb_batch_gpu)
            indices_cpu = cp.asnumpy(indices)
            
            for i in range(len(batch_idx_arr)):
                neighbor_labels = labels_fit[indices_cpu[i]]
                non_noise = neighbor_labels[neighbor_labels != -1]
                if len(non_noise) > 0:
                    labels_full[batch_idx_arr[i]] = np.bincount(non_noise).argmax()
                else:
                    labels_full[batch_idx_arr[i]] = -1
            
            del emb_batch_gpu
            cp.get_default_memory_pool().free_all_blocks()
        
        del emb_fit_gpu, knn, emb_fit, labels_fit
        cp.get_default_memory_pool().free_all_blocks()
        gc.collect()
        
        print(f'   ✅ Assignment complete')

else:
    # ========================================================================
    # CPU CLUSTERING PATH (Fallback)
    # ========================================================================
    MAX_FIT_SAMPLES = 10000000
    
    print(f'\n📊 CPU Dataset Info:')
    print(f'   Total samples: {n_total:,}')
    print(f'   MAX_FIT_SAMPLES: {MAX_FIT_SAMPLES:,}')
    
    if n_total <= MAX_FIT_SAMPLES:
        print(f'\n💻 CPU: Full dataset clustering')
        print(f'   Using ball_tree algorithm')
        print(f'   Expected time: 30-60 minutes')
        
        start_time = time.time()
        model = DBSCAN(eps=CHOSEN_EPS, min_samples=CHOSEN_MIN_SAMPLES,
                      algorithm='ball_tree', n_jobs=-1)
        labels_full = model.fit_predict(emb)
        elapsed = time.time() - start_time
        
        print(f'   ✅ CPU DBSCAN completed in {elapsed/60:.1f} minutes')
        del model
        gc.collect()
    else:
        print(f'\n💻 CPU: Hybrid approach (dataset too large)')
        print(f'   Tip: Use GPU for {n_total:,} samples!')
        
        # Hybrid approach (similar to GPU batch but CPU-based)
        rng = np.random.RandomState(RANDOM_STATE)
        fit_idx = rng.choice(n_total, MAX_FIT_SAMPLES, replace=False)
        emb_fit = emb[np.sort(fit_idx)]
        
        start_time = time.time()
        model = DBSCAN(eps=CHOSEN_EPS, min_samples=CHOSEN_MIN_SAMPLES,
                      algorithm='ball_tree', n_jobs=-1)
        labels_fit = model.fit_predict(emb_fit)
        
        elapsed = time.time() - start_time
        print(f'   ✅ Fit completed in {elapsed/60:.1f} minutes')
        
        # k-NN assignment (CPU)
        remaining_idx = np.setdiff1d(np.arange(n_total), fit_idx)
        labels_full = np.zeros(n_total, dtype=np.int32)
        labels_full[fit_idx] = labels_fit
        
        if len(remaining_idx) > 0:
            print(f'\n   Assigning {len(remaining_idx):,} remaining points...')
            knn = NearestNeighbors(n_neighbors=5, algorithm='ball_tree', n_jobs=-1)
            knn.fit(emb_fit)
            
            BATCH_SIZE = 100000
            n_batches = (len(remaining_idx) + BATCH_SIZE - 1) // BATCH_SIZE
            
            for batch_idx in tqdm(range(n_batches), desc='CPU k-NN', unit='batch'):
                start_idx = batch_idx * BATCH_SIZE
                end_idx = min((batch_idx + 1) * BATCH_SIZE, len(remaining_idx))
                batch_remaining_idx = remaining_idx[start_idx:end_idx]
                
                distances, indices = knn.kneighbors(emb[batch_remaining_idx])
                
                for i in range(len(batch_remaining_idx)):
                    neighbor_labels = labels_fit[indices[i]]
                    non_noise = neighbor_labels[neighbor_labels != -1]
                    if len(non_noise) > 0:
                        labels_full[batch_remaining_idx[i]] = np.bincount(non_noise).argmax()
                    else:
                        labels_full[batch_remaining_idx[i]] = -1
            
            del knn
        
        del model, emb_fit, labels_fit
        gc.collect()

# Save results
print(f'\n💾 Saving results...')
np.save('dbscan_labels.npy', labels_full)
config_array = np.array([
    CHOSEN_EPS, 
    CHOSEN_MIN_SAMPLES, 
    1 if USE_COSINE_DISTANCE else 0,
    1 if USE_GPU else 0
])
np.save('dbscan_config.npy', config_array)

print(f'✅ Saved: dbscan_labels.npy ({len(labels_full):,} labels)')
print(f'✅ Saved: dbscan_config.npy')

# Stats
n_clusters_final = len(set(labels_full) - {-1})
n_noise_final = np.sum(labels_full == -1)
print(f'\n📊 Final Clustering Results:')
print(f'   Total samples: {n_total:,}')
print(f'   Clusters found: {n_clusters_final}')
print(f'   Noise points: {n_noise_final:,} ({n_noise_final/n_total*100:.1f}%)')
print(f'   Clustered points: {n_total - n_noise_final:,} ({(n_total-n_noise_final)/n_total*100:.1f}%)')

gc.collect()
print(f'\n✅ Part 4 complete')
````
